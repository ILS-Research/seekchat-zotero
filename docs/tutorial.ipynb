## 1 Wofür das gut ist

Wer mit Zotero arbeitet, kennt die Routinen, die Zeit kosten, ohne wissenschaftlich zu sein: Literaturlisten abtippen,
Eintragstypen und Jahreszahlen korrigieren, Schlagwörter vereinheitlichen, Sammlungen pflegen, herausfinden, in welchem
der 300 gespeicherten PDFs ein Argument stand.

SeekChat lässt ein **Sprachmodell auf Ihrem eigenen Server** (nicht in einer fremden Cloud) mit Ihrer Bibliothek arbeiten:

- **Fragen an ein PDF** – mit Seitenangaben, die sich anklicken lassen, damit man die Stelle selbst prüft.
- **Aufträge an die Bibliothek** in normaler Sprache – das Modell sucht, liest und schlägt Änderungen vor.
  **Geändert wird erst, nachdem Sie die Vorschau bestätigt haben.**

Das ersetzt nicht das Lesen. Es nimmt Routinearbeit ab und führt schneller zu den Stellen, die man lesen muss.

*Die Bilder stammen aus einem echten Lauf mit einer Beispielbibliothek (deutsche und englische Fachliteratur).*

## 2 Einrichtung

1. Die Dateien (`.xpi`) in Zotero installieren: **Werkzeuge → Add-ons → Zahnrad → Add-on aus Datei installieren**.
   Nötig ist SeekChat; für die Suche nach Bedeutung außerdem ZotSeek, für Bücher SeekBook, für Literaturverzeichnisse Find Online References.
2. **Einstellungen → SeekChat**: Adresse des Modellservers eintragen (Ihr Rechenzentrum oder Institut), **Verbindung testen**, Modell wählen.
3. Der Server bekommt Ihre Fragen und die benötigten Textstellen. Deshalb muss er unter **Erlaubte entfernte Hosts** ausdrücklich eingetragen werden – Sie entscheiden, wohin Texte gehen.
4. **Modellwahl:** Für den Chat über die ganze Bibliothek braucht es ein großes, offenes Modell, das mit Zotero-Funktionen
   umgehen kann (wir nutzen `qwen3.8:27b`, es braucht etwa 40 GB Speicher). Kleinere Modelle reichen für Fragen an ein einzelnes PDF.
   Sprechen Sie das mit der IT ab.

## 2 Einrichtung: Suche nach Bedeutung (ZotSeek)

Die normale Zotero-Suche findet nur Wörter, die tatsächlich im Text stehen. **ZotSeek** legt dazu einmalig einen Index an, der Textstellen
nach ihrer *Bedeutung* ordnet: „Wärmeinseln“ findet dann auch „Hitzeinseln“ und „urban heat island“.

- Einmalig *Update Index* ausführen; bei vielen PDFs dauert das (je nach Server Minuten bis Stunden).
- **AI Agent Access** einschalten – sonst kann SeekChat den Index nicht abfragen.
- Wir lassen die Berechnung auf dem Server laufen (Modell `qwen3-embedding:8b`) statt auf dem eigenen Rechner. Das ist schneller und genauer;
  Voraussetzung ist die angepasste ILS-Version von ZotSeek.

**Wichtig:** Fragen sind erst dann vollständig, wenn der Index fertig ist. Die Einstellungsseite zeigt, wie viele Einträge schon erfasst sind.

<img src="img/settings-zotseek.png" alt="Einstellungen ZotSeek" style="max-width:70%;border:1px solid #ccc">

## 2 Einrichtung: Bücher (SeekBook)

Ein Buch mit 300 Seiten ist etwas anderes als ein Aufsatz. **SeekBook** erfasst Bücher mit Kapiteln und gedruckten Seitenzahlen, damit Treffer
mit Kapitel und Seite zitierbar sind. Bei uns dauerte die Erfassung von vier Büchern knapp drei Minuten.
Im PDF-Chat gibt es einen Knopf, der ein noch nicht erfasstes Dokument dorthin schickt.

# 3 Fragen an ein einzelnes PDF

## 3 Der Chat in der Seitenleiste

Eintrag mit PDF auswählen oder im Reader öffnen: in der Seitenleiste erscheint **„Chat mit PDF“**.

Typische Fragen: *Was ist die Forschungsfrage?* · *Welche Methode wurde verwendet?* · *Was sagt Kapitel 3 zu …?*
Jede Seitenangabe in der Antwort (`[S. 12]`) ist ein Link, der das PDF an dieser Stelle öffnet – **so lässt sich jede Aussage am Original prüfen.**

<img src="img/pdf-2-answer.png" alt="Antwort mit Seitenangaben" style="max-width:70%;border:1px solid #ccc">

## 3 Lange Dokumente, Notizen

Passt ein Text nicht ganz ins Modell, werden die zur Frage passenden Seiten ausgewählt; die Antwort sagt, welche das waren.
Ist das PDF im Reader geöffnet, geht die **aktuelle Seite** mit. Eigene Notizen zum Eintrag lassen sich mitschicken.
Eine Antwort (oder der ganze Chat) kann als **Zotero-Notiz** gespeichert werden. Webseiten-Schnappschüsse, E-Books und Textdateien gehen ebenfalls – ohne Seitenzahlen.

# 4 Der Chat über die ganze Bibliothek

## 4 Das Fenster

Der Knopf mit der Sprechblase und dem Plus in der Zotero-Symbolleiste öffnet ihn. Oben stehen das **Ziel** (ganze Bibliothek oder eine Sammlung) und das Modell.
Darunter die Liste der **Fähigkeiten** (Suchen, Lesen, Importieren, Ändern, Notizen …); einzelne lassen sich abschalten.

## 4 Was das Modell tut – und was Sie bestätigen

- Was das Modell nachschlägt, steht in **einem eingeklappten Block** unter der Antwort. Aufklappen zeigt, welche Suchen gemacht und welche Einträge gelesen wurden.
- **Alles, was Ihre Bibliothek verändert, erscheint vorher als Vorschau mit Häkchen** (alt → neu). Erst nach „Anwenden“ passiert etwas;
  einzelne Änderungen lassen sich abwählen.
- Nur Lesen und Suchen läuft ohne Rückfrage.

# 5 Aufgaben aus der Literaturarbeit

## 5.1 Quellen aus einer Literaturliste importieren

**Anfrage:** *Importiere diese Quellen: Oke 1982 …, Kuttler 2011 …, Wilkinson 2016 …*

Eine eingefügte Liste wird Eintrag für Eintrag aufgelöst (über DOI, ISBN, arXiv-Kennung oder Titel); frei verfügbare PDFs werden mitgeladen. Die Vorschau zeigt, was angelegt würde.

> **Ergebnis:** Zwei von drei Quellen importiert; Wilkinson (2016) war schon vorhanden und wurde nicht doppelt angelegt.

**Zu beachten:** Quellen ohne DOI werden nach Titel gesucht und können falsch zugeordnet werden – Treffer in der Vorschau prüfen.

## 5.2 Die eigene Bibliothek befragen

**Anfrage:** *Welche Einträge behandeln Forschungsdatenmanagement?*

Statt Ordner und Schlagwörter durchzusehen: Das Modell sucht in Titeln, Feldern und Volltexten und nennt die Fundstellen mit Seite. Jeder Eintrag ist verlinkt.

**Zu beachten:** Das Modell sucht in dem, was erfasst ist. PDFs ohne Index und gescannte Dokumente ohne Text bleiben unsichtbar.

**Anfrage:** *Was steht auf der ersten Seite von „Nahversorgung in ländlichen Räumen“?*

Das Modell kann einzelne Seiten eines PDFs (oder Abschnitte einer Webseite) lesen. Das hilft etwa zu klären, was ein Dokument tatsächlich ist, wenn die Metadaten dazu nichts sagen.

## 5.6 Metadaten prüfen und korrigieren

**Anfrage:** *Prüfe, ob der Eintragstyp von „GIS and green infrastructure“ stimmt; ändere nur den Typ.*

Das Modell sieht sich das Titelblatt an, begründet seinen Vorschlag und zeigt die Änderung vor. Felder, die der neue Typ nicht kennt, werden vorher genannt.

> **Ergebnis:** Es ist ein Zeitschriftenartikel (PLOS ONE), kein Buch – der Typ wurde geändert.

**Zu beachten:** Sagen Sie, was geändert werden soll („nur den Typ“). Sonst schlägt das Modell mitunter weitere Felder vor, zum Beispiel das Jahr. Das war im Beispiel nützlich, sollte aber bewusst geschehen.

**Anfrage:** *Welche Schlagwörter nutze ich für Forschungsdaten? Gibt es doppelte Schreibweisen?*

## 5.12 Viele Einträge auf einmal prüfen

**Anfrage:** *Prüfe alle Einträge der Sammlung „Typen prüfen“ auf den richtigen Eintragstyp.*

Bei vielen Einträgen übergibt das Modell die Lesearbeit an einen **Hilfsagenten**, der sie in Paketen zu 20 durchsieht und nur das Ergebnis zurückgibt. Der Chat bleibt übersichtlich. Der Hilfsagent liest nur; geändert wird erst nach Ihrer Bestätigung. Ein eigener Knopf **Subagent stoppen** bricht ihn ab.

> **Ergebnis:** Von fünf Einträgen hatten zwei den falschen Typ: eine Dissertation war als Zeitschriftenartikel erfasst, ein Konferenzbeitrag ebenso.

**Zu beachten:** Bei hunderten Einträgen dauert das Minuten bis Stunden, je nach Server.

## 6 Was man beachten sollte

- **Es ist ein Sprachmodell, kein Rechercheur.** Es kann Dinge falsch zusammenfassen oder Treffer übersehen. Seitenlinks und Vorschauen sind dazu da, an der Quelle nachzusehen.
- **Bestätigen heißt prüfen.** Schauen Sie sich Änderungen in der Vorschau an, besonders bei vielen Einträgen. Zotero kennt keinen „Rückgängig“-Knopf für Metadaten; bei Massenänderungen vorher eine Sicherung ziehen.
- **Klare Aufträge helfen:** „nur den Typ“, „nur in dieser Sammlung“, „höchstens 20 Einträge“.
- **Datenschutz:** Texte gehen nur an Server, die Sie ausdrücklich eingetragen haben – nicht an einen Cloud-Dienst.
- **Modell und Rechenzeit:** Größere Modelle sind verlässlicher, aber langsamer. Aufgaben über viele Einträge dauern.

## 7 Wenn etwas nicht klappt

- **Keine Verbindung:** Adresse stimmt? Server unter *Erlaubte entfernte Hosts* eingetragen? **Verbindung testen**.
- **Das Modell „redet nur“ und tut nichts:** das Modell ist zu klein oder nicht für Zotero-Funktionen geeignet – anderes Modell wählen.
- **Keine Treffer in ZotSeek/SeekBook:** *AI Agent Access* eingeschaltet? Index fertig (Zahlen in den Einstellungen)?
- **Nachsehen, was passiert ist:** Zotero → *Werkzeuge → Entwickler → Browser-Konsole*, Filter `[SeekChat`. Dort steht jeder Schritt mit Zeit.

## 8 Ausblick

- Der Chat über die Bibliothek geht in dem hier gezeigten Chat auf; der einfache PDF-Chat bleibt.
- Geplant: Schlagwörter in der ganzen Bibliothek umbenennen, Volltexte zu vorhandenen Einträgen suchen, Word-/ODT-Dateien lesen.
