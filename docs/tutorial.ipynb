**Was sich bei deiner Arbeit mit der Zotero-Bibliothek abnehmen lässt – und wo du selbst genau hinsehen musst.**
Ein Tutorial für ILS-Kolleginnen und -Kollegen. Das Modell läuft auf unserer GPU-Maschine im Serverraum, nicht in einer fremden Cloud.

## 1 Wofür das gut ist

Routinen, die Zeit kosten, ohne wissenschaftlich zu sein: Literaturlisten abtippen, Eintragstypen und Jahreszahlen korrigieren,
Schlagwörter vereinheitlichen, Sammlungen pflegen, herausfinden, in welchem von 300 PDFs ein Argument stand.

SeekChat lässt ein **Sprachmodell auf dem ILS-Server** mit deiner Bibliothek arbeiten. Zwei Wege:

- **Fragen an ein PDF** (Kapitel 3) – die Antwort nennt Seitenzahlen, die du anklicken kannst, um die Stelle selbst zu prüfen.
- **Aufträge an die ganze Bibliothek** (Kapitel 4 und 5) – in normaler Sprache; das Modell sucht, liest und schlägt Änderungen vor.
  **Geändert wird erst, nachdem du die Vorschau bestätigt hast.**

Das ersetzt nicht das Lesen. Es nimmt dir Routine ab und führt dich schneller zu den Stellen, die du lesen musst.

*Alle Bilder stammen aus einem echten Lauf mit einer Beispielbibliothek (deutsche und englische Fachliteratur).*

## 2 Einrichtung: Plugins und Modellserver

**Plugins installieren.** Alle Dateien liegen auf dem ILS-Download-Portal (`zotero.ils.local/downloads`).

1. Die `.xpi`-Dateien laden: **SeekChat** (immer), `ZotSeek` (Suche nach Bedeutung), `SeekBook` (Bücher und Dokumente über 30 Seiten),
   `Find Online References` (Quellen aus Literaturverzeichnissen übernehmen, bessere Titelsuche beim Import).
2. In Zotero: **Werkzeuge → Plugins → Zahnrad → Plugin aus Datei installieren…**, Datei wählen, Zotero neu starten. Updates kommen danach automatisch über das Portal.

**Modellserver eintragen.** Unser Server ist `https://ollama.ils.local`, die GPU-Maschine im Serverraum; er ist nur im ILS-Netz erreichbar.

1. **Einstellungen → SeekChat**, Schnittstelle **Ollama**, Serveradresse `https://ollama.ils.local`.
2. **Erlaubte entfernte Hosts:** `ollama.ils.local` eintragen. SeekChat schickt Texte nur an Server, die du hier ausdrücklich nennst – ohne den Eintrag passiert nichts.
3. **Verbindung testen** – bei Erfolg erscheint die Modellliste.
4. Modell **`qwen3_8_27b_128k`** wählen. Es ist das Modell, mit dem alle Beispiele in diesem Tutorial entstanden sind. Für den Chat über die ganze Bibliothek (ab Kapitel 4) braucht es ein starkes Modell; kleinere reichen nur für Fragen an ein einzelnes PDF.

## 2 Einrichtung: ZotSeek – Suche nach Bedeutung

Die normale Zotero-Suche findet nur Wörter, die im Text stehen. **ZotSeek** legt einmalig einen Index an, der Textstellen nach ihrer *Bedeutung* ordnet: „Wärmeinseln“ findet dann auch „Hitzeinseln“ und „urban heat island“.

1. **Einstellungen → ZotSeek**. Als Einbettungsmodell das Servermodell **`qwen3-embedding:8b`** wählen. Es läuft auf derselben GPU-Maschine; das geht nur mit der ILS-Version von ZotSeek vom Portal.
2. **AI Agent Access** einschalten – sonst kann SeekChat den Index nicht abfragen.
3. **Update Index** drücken und warten. Bei vielen PDFs dauert das je nach Auslastung des Servers Minuten bis Stunden; du kannst in der Zeit weiterarbeiten.
4. Oben in den Einstellungen siehst du, wie viele Einträge schon erfasst sind.

**Wichtig:** Solange der Index nicht fertig ist, bleiben nicht erfasste PDFs für die Suche unsichtbar (siehe 5.2 und 5.3).

Achte auf: „Papers Indexed“ (wie viele Einträge schon im Index sind) und das Modell in der Zeile darunter (hier das Servermodell qwen3-embedding-8b).</em></figcaption></figure>

## 2 Einrichtung: SeekBook – Bücher

Ein Buch mit 300 Seiten ist etwas anderes als ein Aufsatz. **SeekBook** erfasst Bücher mit Kapiteln und gedruckten Seitenzahlen, damit Treffer mit Kapitel und Seite zitierbar sind.

1. **Einstellungen → SeekBook**: dasselbe Servermodell `qwen3-embedding:8b` eintragen.
2. Bücher in Zotero markieren und von SeekBook indexieren lassen. Bei uns dauerte es für vier Bücher knapp drei Minuten.
3. Alternativ gibt es im PDF-Chat (Kapitel 3) einen Knopf, der ein noch nicht erfasstes Buch an SeekBook übergibt.

# 3 Fragen an ein einzelnes PDF

## 3 Der Chat in der Seitenleiste

1. Eintrag mit PDF in Zotero auswählen (oder das PDF im Reader öffnen).
2. In der rechten Seitenleiste den Abschnitt **„Chat mit PDF“** suchen und, falls zugeklappt, aufklappen.
3. Frage ins Feld schreiben, **Enter** sendet (Shift+Enter = neue Zeile).

Typische Fragen: *Was ist die Forschungsfrage?* · *Welche Methode wurde verwendet?* · *Was sagt Kapitel 3 zu …?*
Jede Seitenangabe in der Antwort (z. B. `[S. 12]`) ist ein Link, der das PDF an dieser Stelle öffnet – **so prüfst du jede Aussage am Original.**

## 3 Lange Dokumente

Passt ein Text nicht ganz ins Modell, wählt SeekChat die zur Frage passenden Seiten aus; die Antwort sagt dir, welche das waren.
Bücher mit eigenem Index (SeekBook) werden gezielt durchsucht – dafür muss der Index fertig sein (Kapitel 2).

- Ist das PDF im **Reader** geöffnet, geht die **aktuelle Seite** mit („erkläre mir diese Seite“).
- Eigene **Notizen** zum Eintrag lassen sich per Häkchen („Notizen als Kontext“) mitschicken.
- Webseiten-Schnappschüsse, E-Books und Textdateien gehen ebenfalls, aber ohne Seitenzahlen.
- Eine inhaltlich anspruchsvolle Frage an ein ganzes Buch findest du in 5.14.
- Was mit dem Verlauf passiert und wie du ihn sicherst: Kapitel 6.

# 4 Der Chat über die ganze Bibliothek

## 4 Das Fenster öffnen

1. In der Zotero-Symbolleiste den Knopf mit der **Sprechblase und dem Plus** drücken.
2. Oben das **Ziel** wählen: die ganze Bibliothek oder eine Sammlung (vorgewählt ist, was in Zotero gerade offen ist).
3. Darunter steht die Liste der **Fähigkeiten** (Suchen, Lesen, Importieren, Ändern, Notizen …). Klapp sie auf, wenn du einzelne abschalten willst.
   Ausgegraut sind Fähigkeiten, für die ein Plugin fehlt (Kapitel 2).

## 4 Was das Modell tut – und was du bestätigst

- Was das Modell nachschlägt, steht in **einem eingeklappten Block** unter der Antwort („7 Werkzeugaufrufe …“). Aufklappen zeigt jede Suche und jeden gelesenen Eintrag.
- **Alles, was deine Bibliothek verändert, erscheint vorher als Vorschau mit Häkchen** (alt → neu). Erst nach **Anwenden** passiert etwas; einzelne Änderungen kannst du abwählen.
- Reines Suchen und Lesen läuft ohne Rückfrage.
- Läuft etwas zu lange: den **Stopp**-Knopf neben dem Senden-Knopf drücken.

<figure style="margin:0.6em 0"><img src="img/fix-one-1-confirm-update_item.png" alt="Vorschau mit Bestätigung" style="max-width:75%;border:1px solid #ccc"><figcaption style="font-size:0.75em;color:#555;max-width:75%"><em>Vorschau einer Änderung. # 5 Aufgaben aus der Literaturarbeit

## 5.1 Quellen aus einer Literaturliste importieren

**Du schreibst:** *Importiere diese Quellen: Oke 1982 …, Kuttler 2011 …, Wilkinson 2016 …*

Füge die Literaturliste ins Feld ein. Jeder Eintrag wird aufgelöst (über DOI, ISBN, arXiv-Kennung oder Titel); frei verfügbare PDFs werden mitgeladen. Prüfe die Vorschau, bevor du bestätigst.

> **Ergebnis:** Zwei von drei Quellen importiert; Wilkinson (2016) war schon vorhanden und wurde nicht doppelt angelegt.

**Zu beachten:** Quellen ohne DOI werden über den Titel gesucht und können falsch zugeordnet werden – vergleiche die Titel in der Vorschau. Mit `Find Online References` (Kapitel 2) wird die Titelsuche besser.

## 5.2 Die eigene Bibliothek befragen

**Du schreibst:** *Welche Einträge behandeln Forschungsdatenmanagement? Nenne sie mit Jahr.*

Statt Ordner und Schlagwörter durchzusehen: Das Modell sucht in Titeln, Feldern und Volltexten und nennt Fundstellen mit Seite. Jeder Eintrag ist verlinkt.

**Zu beachten:** Das Modell findet nur, was erfasst ist. PDFs ohne Index und gescannte Dokumente ohne Textebene bleiben unsichtbar – lege die Indizes an (Kapitel 2: ZotSeek und SeekBook).

## 5.3 Passagen finden, auch ohne die richtigen Stichwörter

**Du tust:** zur Frage aus 5.2 den eingeklappten Block aufklappen.

So siehst du, woher ein Treffer kam: normale Zotero-Suche, Suche nach Bedeutung (ZotSeek) oder Buchsuche (SeekBook, mit Kapitel und Seite). Der Auszug lässt dich eine Stelle beurteilen, bevor du das PDF öffnest.

**Zu beachten:** Steht bei einem Treffer nur „Zotero“, obwohl du Volltext erwartest, ist der Index vermutlich nicht fertig (Kapitel 2).

## 5.6 Metadaten prüfen und korrigieren

**Du schreibst:** *Prüfe, ob der Eintragstyp von „GIS and green infrastructure“ stimmt; ändere nur den Typ.*

Das Modell sieht sich das Titelblatt an, begründet seinen Vorschlag und zeigt die Änderung vor. Felder, die der neue Typ nicht kennt, werden vorher genannt.

> **Ergebnis:** Es ist ein Zeitschriftenartikel (PLOS ONE), kein Buch – der Typ wurde geändert.

**Zu beachten:** Schreib, was geändert werden soll („nur den Typ“). Sonst schlägt das Modell mitunter weitere Felder vor, etwa das Jahr. Im Beispiel war das nützlich, sollte aber bewusst passieren.

## 5.12 Viele Einträge auf einmal prüfen

**Du schreibst:** *Prüfe alle Einträge der Sammlung „Typen prüfen“ auf den richtigen Eintragstyp.*

Bei vielen Einträgen übergibt das Modell die Lesearbeit an einen **Hilfsagenten**. Er geht sie in Paketen zu 20 durch, liest nur und gibt am Ende ein Ergebnis zurück; der Chat bleibt übersichtlich. Geändert wird erst nach deiner Bestätigung. Ein eigener Knopf **Subagent stoppen** bricht nur ihn ab; was er bis dahin hatte, bleibt.

> **Ergebnis:** Von fünf Einträgen hatten zwei den falschen Typ: eine Dissertation war als Zeitschriftenartikel erfasst, ein Konferenzbeitrag ebenso.

**Zu beachten:** Bei hunderten Einträgen dauert das Minuten bis Stunden, je nach Auslastung des Servers.

## 5.14 Ein Buch inhaltlich befragen

**Du schreibst (PDF-Chat, Kapitel 3):** *Welche Phasen unterscheidet das Buch im Lebenszyklus von Forschungsdaten, und was verlangt es in der Planungsphase konkret von Forschenden? Nenne die Stellen mit Seitenangabe.*

Bisher haben die Beispiele Zotero-Aufgaben in Sprache übersetzt. Hier geht es um die Literatur selbst: Du willst ein **Vorgehen aus einem langen Buch herausarbeiten** – die Phasen eines Modells, und was daraus für dein eigenes Projekt folgt. Das ist eine Frage an *ein* Buch (Schlenz et al., 109 Seiten), deshalb der PDF-Chat in der Seitenleiste: Er liest das Buch ganz und belegt mit Seiten.

1. Eintrag auswählen, im Abschnitt „Chat mit PDF“ die Frage stellen. Formuliere **zwei Teilfragen** und verlange **Seitenangaben**.
2. Die Zeile über der Antwort zeigt, was gelesen wurde (hier: vollständiger Text, 109 Seiten).
3. **Klick auf die Seitenlinks**, bevor du etwas übernimmst: Steht an der Stelle, was die Antwort behauptet?
4. Hake die Frage nicht ab, wenn nur der erste Teil beantwortet ist: Frag nach (*„Und was verlangt das Buch in der Planungsphase konkret?“*).

> **Ergebnis (Beginn der Antwort):** Das Buch beschreibt den Lebenszyklus nicht als feste Liste von Phasen, sondern verweist auf einen typischen Lebenszyklus aus Erfassung, Prozessierung/Datenanalyse, Publikation, Archivierung und Teilen von Daten [S. 11].

## 5.15 Zwei Bücher vergleichen

**Du schreibst:** *Vergleiche, was „Forschungsdatenmanagement – eine praxisorientierte Einführung“ und „Forschungsdaten-Policies für Forschungsprojekte“ dazu sagen, wie ein Projekt seine Regeln zum Umgang mit Forschungsdaten festlegen soll. Wo ergänzen sich die Bücher, wo setzen sie unterschiedliche Schwerpunkte? Belege jede Aussage mit Buch und Seite.*

Das ist eine Frage, wie du sie beim Schreiben eines Forschungsstands stellst. Das Modell sucht in beiden Büchern (SeekBook, Kapitel 2), liest die Passagen nach und ordnet sie. Der Lauf dauerte hier einige Minuten (mehrere Such- und Leseschritte); die Zwischenschritte stehen im eingeklappten Block.

> **Ergebnis (gekürzt):** Die beiden Bücher behandeln dieselbe Frage auf zwei Ebenen. Schmiederer & Kuberek (2022) beschreiben, *wie* ein Projekt zu einer verbindlichen Forschungsdaten-Policy kommt (Beteiligte, vier Phasen, Verabschiedung, Normenhierarchie). Schlenz et al. (2026) zeigen, *was* in den Regeln stehen muss, und behandeln den Datenmanagementplan als zentrales Dokument. Bei den Inhalten ergänzen sie sich am stärksten; bei der Frage, ob eine eigene Policy nötig ist, setzen sie verschiedene Schwerpunkte.

**So liest du das Ergebnis:**

1. Jede Aussage hat Buch und Seite. **Schlag mindestens die Stellen auf, die du zitieren willst** – das Modell kann Passagen falsch zuordnen oder verkürzen.
2. Prüfe die Gegenüberstellung: Hat das Modell wirklich *beide* Bücher zu jedem Punkt herangezogen, oder steht bei einem Punkt nur ein Buch?
3. Aussagen wie „Buch X sagt dazu nichts“ sind der heikelste Teil: Das Modell hat nur gelesen, was die Suche ihm zeigte. Frag nach (*„Suche im Buch X gezielt nach …“*), bevor du so etwas übernimmst.

**Zu beachten:** Die Antwort ist ein Arbeitsentwurf für deinen eigenen Text, keine Quelle. Sichere sie als Notiz (Kapitel 6), wenn du damit weiterarbeiten willst.

# 6 Chats sichern

## 6 Was flüchtig ist und was bleibt

**Chats sind flüchtig.** Ein Chat liegt nur im Arbeitsspeicher von Zotero:

- Er bleibt erhalten, solange Zotero läuft – auch wenn du das Fenster schließt oder zu einem anderen Eintrag wechselst und zurückkommst. Beim PDF-Chat gibt es einen Chat **pro Dokument**.
- Er ist weg nach **„Neuer Chat“** und nach einem **Neustart von Zotero** (auch bei einem Absturz oder einem Update).

**PDF-Chat (Seitenleiste):**

- **Verlauf als Notiz** speichert das ganze Gespräch als Zotero-Notiz am Eintrag; die Seitenangaben bleiben anklickbare Links.
- Der Knopf **„Als Notiz“ an einer einzelnen Antwort** übernimmt nur diese Antwort.
- Der Verlauf lässt sich außerdem als Markdown-Datei sichern.

**Chat über die Bibliothek (Kapitel 4):** Dieses Fenster hat derzeit **keinen Speichern-Knopf**. Was der Chat *getan* hat, bleibt in Zotero (importierte Einträge, geänderte Metadaten, Sammlungen, Notizen). Der **Text des Gesprächs** selbst geht verloren. Zwei Auswege:

1. Bitte das Modell um eine Notiz: *Schreibe die Antwort als Notiz an den Eintrag „…“* (siehe 5.9).
2. Markiere den Text im Fenster, kopiere ihn und füge ihn in ein Dokument oder eine Notiz ein.

Ein Export für dieses Fenster ist geplant.

## 7 Was du beachten solltest

- **Es ist ein Sprachmodell, kein Rechercheur.** Es kann Dinge falsch zusammenfassen oder Treffer übersehen. Seitenlinks und Vorschauen sind dazu da, an der Quelle nachzusehen (5.14, 5.15).
- **Bestätigen heißt prüfen.** Sieh dir Änderungen in der Vorschau an, besonders bei vielen Einträgen. Zotero kennt für Metadaten keinen „Rückgängig“-Knopf; bei Massenänderungen vorher eine Sicherung ziehen.
- **Klare Aufträge helfen:** „nur den Typ“ (5.6), „nur in dieser Sammlung“, „höchstens 20 Einträge“.
- **Ohne Index keine Treffer:** PDFs, die ZotSeek oder SeekBook noch nicht erfasst haben, bleiben bei der Suche unsichtbar (Kapitel 2).
- **Datenschutz:** Texte gehen nur an `ollama.ils.local`, den Server im Serverraum, und nur, wenn er unter „Erlaubte entfernte Hosts“ eingetragen ist – nicht an einen Cloud-Dienst.
- **Rechenzeit:** Der Server wird von mehreren geteilt. Aufgaben über viele Einträge oder ganze Bücher dauern Minuten; währenddessen kannst du weiterarbeiten.
- **Chats sind flüchtig** (Kapitel 6): Wichtiges als Notiz sichern.

## 8 Wenn etwas nicht klappt

- **Keine Verbindung:** Bist du im ILS-Netz? Stimmt die Adresse `https://ollama.ils.local`? Steht `ollama.ils.local` unter *Erlaubte entfernte Hosts*? Dann **Verbindung testen** (Kapitel 2).
- **Das Modell „redet nur“ und tut nichts:** Das gewählte Modell ist zu klein oder nicht für Zotero-Funktionen geeignet – wähle `qwen3_8_27b_128k`.
- **Keine Treffer aus ZotSeek oder SeekBook:** Ist *AI Agent Access* eingeschaltet? Ist der Index fertig (Zahlen in den Einstellungen, Kapitel 2)?
- **Antwort bricht ab oder dauert sehr lange:** Der Server ist ausgelastet. Mit dem Stopp-Knopf abbrechen und es später oder mit einer kleineren Frage (weniger Einträge, weniger Seiten) noch einmal versuchen.
- **Chat weg:** Siehe Kapitel 6 – er war nur im Arbeitsspeicher.
- **Nachsehen, was passiert ist:** Zotero → *Werkzeuge → Entwickler → Browser-Konsole*, Filter `[SeekChat`. Dort steht jeder Schritt mit Zeit; das hilft der IT bei der Fehlersuche.

## 9 Ausblick

- Der Chat über die Bibliothek (Kapitel 4) soll den älteren Bibliotheks-Chat ablösen; der einfache PDF-Chat (Kapitel 3) bleibt.
- Geplant: Export und Speichern für den Chat über die Bibliothek (Kapitel 6), Schlagwörter in der ganzen Bibliothek umbenennen, Volltexte zu vorhandenen Einträgen suchen, Word- und ODT-Dateien lesen.
