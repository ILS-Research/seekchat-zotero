# SeekChat – Tutorial: PDF-Chat und Werkzeug-Chat

**Fragen an ein PDF stellen und in normaler Sprache sagen, was in Zotero passieren soll – mit einem Sprachmodell, das im Haus läuft.**

## Inhalt

1. **Worum es geht** – PDF-Chat und Werkzeug-Chat, dazu die Helfer: ZotSeek, SeekBook, Find Online References
2. **Installation und Einrichtung** – Plugins installieren, Ollama-Server eintragen, ein werkzeugfähiges Modell wählen
3. **Der einfache PDF-Chat** – Fragen an ein einzelnes Dokument in der Seitenleiste, Antworten mit anklickbaren
   Seitenangaben, lange Dokumente und Kapitelauswahl, Webseiten und E-Books, Chat als Notiz speichern
4. **Das Fenster des Werkzeug-Chats** – Ziel (Sammlung/Bibliothek), Werkzeugliste, eingeklappte Werkzeugaufrufe, Vorschau und Bestätigung
5. **Anwendungsfälle**
   1. Quellen aus Text importieren – Literaturliste einfügen, Dubletten erkennen, PDFs mitnehmen
   2. Die Bibliothek befragen – „Welche Paper zu Hitzeinseln habe ich seit 2020?“
   3. Inhalte finden – Suche nach Bedeutung (ZotSeek) und in Büchern bis auf die Seite (SeekBook)
   4. Was ist ausgewählt? – mit markierten Einträgen, der offenen Sammlung oder markiertem Text arbeiten
   5. Dokumente lesen – Titelblatt, Zusammenfassung, bestimmte Seiten eines PDFs, einer Webseite oder eines E-Books
   6. Einträge prüfen und korrigieren – falscher Eintragstyp, fehlende DOI, Jahr, Verlag
   7. Schlagwörter aufräumen – vorhandene Schlagwörter nutzen, ergänzen, entfernen
   8. Sammlungen anlegen und füllen – „Lege alles zu Starkregen in Projekt A / Hitze ab“
   9. Notizen schreiben – Zusammenfassung eines Papers oder Rechercheergebnis als Notiz
   10. Literaturlisten auswerten – passende Quellen aus der Bibliographie eines Papers finden und verlinken
   11. Recherche weiterführen – verwandte Quellen prüfen, ob sie schon in der Bibliothek sind
   12. Große Aufgaben an den Subagenten geben – eine ganze Sammlung durchgehen und Korrekturen vorschlagen
   13. Aufgaben verketten – suchen → lesen → korrigieren → in Sammlung ablegen → Notiz schreiben
6. **Gut zu wissen** – Modellwahl (z. B. qwen3.8:27b, ~40 GB), Datenschutz, nichts ändert sich ohne Bestätigung, Grenzen
7. **Wenn etwas nicht klappt** – Verbindung, Modell ohne Werkzeuge, Protokoll in der Browser-Konsole
8. **Ausblick** – der Bibliotheks-Chat geht im Werkzeug-Chat auf; weitere Werkzeuge in Planung


## 1 Worum es geht

**Zwei Chats, ein Modell, das im Haus läuft:**

- **PDF-Chat** – Fragen an das Dokument in der Seitenleiste, Antworten mit anklickbaren Seitenangaben.
- **Werkzeug-Chat** – das Modell ruft Zotero-Werkzeuge auf: suchen, lesen, ändern, ablegen, Notizen schreiben.

Helfer (optional): **ZotSeek** (Suche nach Bedeutung in Papern), **SeekBook** (Bücher bis auf die Seite),
**Find Online References** (Literaturlisten, bessere Titelsuche beim Import).

*Alle Bilder stammen aus einem echten Lauf (Zotero 10, qwen3.8:27b auf ollama.ils.local) mit echten Dokumenten.*

## 2 Installation und Einrichtung

1. `.xpi`-Dateien laden: Zotero → **Werkzeuge → Add-ons** → Zahnrad → *Add-on aus Datei installieren*.
   SeekChat, dazu nach Bedarf ZotSeek, SeekBook, Find Online References.
2. **Einstellungen → SeekChat**: Schnittstelle *Ollama*, Serveradresse, **Verbindung testen**, Modell wählen.
3. Server außerhalb des eigenen Rechners unter **Erlaubte entfernte Hosts** eintragen – er bekommt dann Fragen und Dokumenttext.
4. Für den Werkzeug-Chat: ein **starkes Open-Weights-Modell mit nativem Tool-Calling**, z. B. `qwen3.8:27b` (~40 GB RAM/VRAM).

<img src="img/settings-seekchat.png" alt="Einstellungen SeekChat" style="max-width:70%;border:1px solid #ccc">

## 2 Einrichtung: ZotSeek mit Servermodell

ZotSeek rechnet Einbettungen normalerweise lokal auf der CPU. Wir tragen stattdessen ein Modell vom Server ein:
**`qwen3-embedding:8b`** (4096 Dimensionen). Das ILS-Fork von ZotSeek kennt dafür *Servermodelle*; der Host muss in den
erlaubten Hosts stehen. Außerdem **AI Agent Access** einschalten – sonst kann SeekChat ZotSeek nicht abfragen.
Dann **Update Index** (oder *Rebuild Index* nach einem Modellwechsel).

<img src="img/settings-zotseek.png" alt="Einstellungen ZotSeek" style="max-width:70%;border:1px solid #ccc">

## 2 Einrichtung: SeekBook

SeekBook indexiert Bücher mit demselben Servermodell. Die Indizes sollten **fertig sein, bevor man fragt** –
im Test dauerte der Index für vier Bücher (zusammen mehrere hundert Seiten) knapp drei Minuten.
Ein Knopf im PDF-Chat reicht ein noch nicht indexiertes Dokument an den richtigen Index weiter.

# 3 Der einfache PDF-Chat

## 3 PDF-Chat in der Seitenleiste

Eintrag mit PDF auswählen (oder im Reader öffnen): die Seitenleiste zeigt **„Chat mit PDF“**.
Frage eintippen – Antwort mit Quellen. Jede Seitenangabe `[S. 12]` ist ein Link, der das PDF dort öffnet.

## 3 Lange Dokumente

Passt das PDF nicht ins Modell, wählt SeekChat die Seiten, die zur Frage passen, und nennt sie.
Im Reader geht die **aktuelle Seite** immer mit; angehakte Notizen des Eintrags ebenfalls.
Antworten und der ganze Chat lassen sich als **Notiz** oder Markdown sichern. Dasselbe gilt für Webseiten-Snapshots, E-Books und Textdateien (ohne Seitenzahlen).

# 4 Das Fenster des Werkzeug-Chats

## 4 Das Fenster

Der Sprechblasen-Knopf mit Plus in der Werkzeugleiste öffnet den Chat. Oben: **Ziel** (Bibliothek oder Sammlung) und Modell.
Die **Werkzeugliste** ist einklappbar und scrollbar; einzelne Werkzeuge lassen sich abschalten, Werkzeuge fehlender Plugins sind grau.

<img src="img/window-1-empty.png" alt="Leeres Chatfenster" style="max-width:60%;border:1px solid #ccc">

<img src="img/window-2-tool-list.png" alt="Werkzeugliste" style="max-width:60%;border:1px solid #ccc">

## 4 Werkzeugaufrufe und Bestätigung

- Alle Aufrufe einer Antwort stehen in **einem eingeklappten Block** mit einer Statuszeile.
- **Nichts ändert sich ohne dich:** jedes ändernde Werkzeug zeigt vorher eine **Vorschau mit Häkchen** und wartet.
  Der Block klappt dann von selbst auf.
- Lesende Werkzeuge laufen ohne Rückfrage.

## 6 Gut zu wissen

- **Modell:** stark, offen, mit nativem Tool-Calling (z. B. `qwen3.8:27b`, ~40 GB). Kleinere Modelle rufen Werkzeuge unzuverlässig auf.
- **Datenschutz:** Dokumente gehen nur an Server in der Liste *Erlaubte entfernte Hosts*.
- **Nichts ohne Bestätigung:** Änderungen immer mit Vorschau.
- **Grenzen:** Das Modell kann zu viel suchen oder zusätzliche Felder ändern – die Vorschau zeigt es; Anweisungen wie „ändere nur den Typ“ helfen.
- Lange Läufe: Der Stopp-Knopf bricht ab; das Ergebnis bis dahin bleibt.

## 7 Wenn etwas nicht klappt

- **Keine Verbindung:** Adresse, *Erlaubte entfernte Hosts*, **Verbindung testen**.
- **Werkzeuge werden nicht genutzt / Text statt Aufruf:** Modell ohne natives Tool-Calling.
- **ZotSeek/SeekBook ohne Treffer:** *AI Agent Access* bzw. Server an? Index fertig (Statistik in den Einstellungen)?
- **Protokoll:** Zotero → *Werkzeuge → Entwickler → Browser-Konsole*, Filter `[SeekChat`.

## 8 Ausblick

- Der Bibliotheks-Chat geht im Werkzeug-Chat auf.
- Geplant: Import direkt aus dem Chat erweitern, Schlagwörter bibliotheksweit umbenennen, Volltexte zu vorhandenen Einträgen finden, Word/ODT.
