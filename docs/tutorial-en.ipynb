## 1 What it is good for

Routines that cost time without being scholarly: retyping reference lists, correcting item types and years,
unifying tags, tidying collections, finding out in which of 300 PDFs an argument appeared.

SeekChat lets a **language model on the ILS server** work with your library. Two ways:

- **Questions to a PDF** (chapter 3) – the answer gives page numbers you can click to check the passage yourself.
- **Requests to the whole library** (chapters 4 and 5) – in plain language; the model searches, reads and proposes changes.
  **Nothing is changed until you have confirmed the preview.**

This does not replace reading. It takes routine off your hands and leads you faster to the passages you have to read.

*All pictures come from a real run with a sample library (German and English specialist literature).*

## 2 Setup: plugins and model server

**Install the plugins.** All files are on the ILS download portal (`zotero.ils.local/downloads`).

1. Download the `.xpi` files: **SeekChat** (always), `ZotSeek` (search by meaning), `SeekBook` (books and documents over 30 pages),
   `Find Online References` (take over sources from bibliographies, better title search when importing).
2. In Zotero: **Tools → Plugins → gear icon → Install Plugin From File…**, choose the file, restart Zotero. Updates then arrive automatically through the portal.

**Enter the model server.** Our server is `https://ollama.ils.local`, the GPU machine in the server room; it can only be reached from the ILS network.

1. **Settings → SeekChat**, interface **Ollama**, server address `https://ollama.ils.local`.
2. **Allowed remote hosts** (“Erlaubte entfernte Hosts”): enter `ollama.ils.local`. SeekChat sends text only to servers you name here explicitly – without the entry nothing happens.
3. **Test connection** (“Verbindung testen”) – on success the model list appears.
4. Choose the model **`qwen3_8_27b_128k`**. All examples in this tutorial were made with it. The chat over the whole library (from chapter 4 on) needs a strong model; smaller ones only suffice for questions to a single PDF.

## 2 Setup: ZotSeek – search by meaning

Zotero's normal search only finds words that are in the text. **ZotSeek** builds an index once that orders passages by their *meaning*: “Wärmeinseln” then also finds “Hitzeinseln” and “urban heat island”.

1. **Settings → ZotSeek**. Choose the server model **`qwen3-embedding:8b`** as embedding model. It runs on the same GPU machine; this only works with the ILS version of ZotSeek from the portal.
2. Switch on **AI Agent Access** – otherwise SeekChat cannot query the index.
3. Press **Update Index** and wait. With many PDFs this takes minutes to hours, depending on how busy the server is; you can keep working meanwhile.
4. At the top of the settings you see how many items are already indexed.

**Important:** As long as the index is not finished, PDFs that are not yet indexed stay invisible to the search (see 5.2 and 5.3).

Look at: “Papers Indexed” (how many items are already in the index) and the model in the line below (here the server model qwen3-embedding-8b).</em></figcaption></figure>

## 2 Setup: SeekBook – books

A 300-page book is something different from an article. **SeekBook** indexes books with chapters and printed page numbers, so that hits can be cited with chapter and page.

1. **Settings → SeekBook**: enter the same server model `qwen3-embedding:8b`.
2. Select books in Zotero and let SeekBook index them. For us, four books took just under three minutes.
3. Alternatively, the PDF chat (chapter 3) has a button that hands a book that is not yet indexed over to SeekBook.

# 3 Questions to a single PDF

## 3 The chat in the side pane

1. Select an item with a PDF in Zotero (or open the PDF in the reader).
2. In the right-hand side pane find the section **“Chat mit PDF”** (Chat with PDF) and expand it if it is collapsed.
3. Type your question into the field; **Enter** sends it (Shift+Enter = new line).

Typical questions: *What is the research question?* · *Which method was used?* · *What does chapter 3 say about …?*
Every page reference in the answer (e.g. `[S. 12]`, “S.” = page) is a link that opens the PDF at that spot – **this is how you check every statement against the original.**

## 3 Long documents

If a text does not fit into the model completely (here: a dissertation of 207 pages), a box **“Dokument zu groß für den Kontext”** (document too large for the context) appears when you open the chat. Only part of it can be read per question; you choose how SeekChat picks the pages:

- **Semantic search (SeekBook / ZotSeek):** the best pages come from the index. This needs a finished index (chapter 2); otherwise the option is greyed out, and a button **“Eintrag an ZotSeek übergeben”** (hand the item over to ZotSeek) sends the document there.
- **Keyword expansion by the model:** the model creates search terms for the question (synonyms, technical terms) in the language of the document; the pages with the most hits go along.
- **Search only in selected chapters:** in the document's table of contents you pick the chapters that come into question.

Above the answer you then see **which pages were read** (“Auszüge: S. 1–2, 13–44 … von 207 Seiten”, i.e. excerpts: pp. …) and with which search terms. For long documents too: click the page links and check.

Further notes:

- If the PDF is open in the **reader**, the **current page** goes along (“explain this page to me”).
- Your own **notes** on the item can be sent along by ticking them (“Notizen als Kontext”, notes as context).
- Web page snapshots, e-books and text files work too, but without page numbers.
- A substantive question to a whole book is in 5.14; what happens to the chat and how to save it: chapter 6.

# 4 The chat over the whole library

## 4 What the model does – and what you confirm

- What the model looks up is in **one collapsed block** below the answer (“7 Werkzeugaufrufe …” = 7 tool calls). Expanding it shows every search and every item read.
- **Everything that changes your library appears first as a preview with ticks** (old → new). Only after you press **apply** does anything happen; you can untick single changes.
- Plain searching and reading runs without asking.
- If something takes too long: press the **Stopp** (stop) button next to the send button.

# 5 Tasks from everyday literature work

## 5.1 Import sources from a reference list

**You write:** *Import these sources: Oke 1982 …, Kuttler 2011 …, Wilkinson 2016 …*

Paste the reference list into the field. SeekChat tries to resolve every entry (via DOI, ISBN, arXiv id or title); freely available PDFs are fetched along. The preview states per source **where the data came from**, and you tick what should be imported. Check it before you confirm.

> **Result:** Two of three sources imported; Wilkinson (2016) was already in the library and was not created twice.

**Keep in mind:** Only sources with a DOI or another identifier are really looked up (“gefunden” = found). For the others the entry is created **from the reference text alone** (“Text”: not looked up) – check authors, title and year yourself. With `Find Online References` (chapter 2) the title search gets better.

## 5.2 Ask your own library

**You write:** *Which items deal with research data management? List them with year.*

Instead of going through folders and tags: the model searches titles, fields and full texts and names the places where it found something, with page. Every item is linked.

**Keep in mind:** The model only finds what is indexed. PDFs without an index and scanned documents without a text layer stay invisible – build the indexes (chapter 2: ZotSeek and SeekBook).

*See also: 5.3 (where a hit came from).*

## 5.6 Check and correct metadata

**You write:** *The item “GIS and green infrastructure” was entered as a book. Check on the first page whether the type is right, and correct it only if necessary. Change nothing else.*

The model reads the first page, justifies its proposal and shows the change beforehand as a preview. Fields that the new type does not know are named there.

> **Result:** It is a journal article (PLOS ONE), not a book – the type was changed.

**Keep in mind:** Despite “change nothing else” the model proposes **more** than the type here: the full title, DOI, issue, pages, journal and volume, each read off the first page. That is factually useful here, but it is not what you asked for. **This is exactly why the preview exists:** untick in the list what you do not want, and only then confirm. Also look at the box “0 Einträge ändern – fehlgeschlagen” (0 items changed – failed) in the block: the model's first attempt failed and it made a second one itself; nothing was changed before you confirmed.

## 5.12 Check many items at once

**You write:** *In the collection “Typen prüfen”, check for every item on the first pages whether the item type is right. Let the subagent do that and afterwards change only the types that are wrong.*

With many items the model hands the reading over to a **helper agent** (called “Subagent” in the program). It goes through them in packages of 20, only reads, and returns a result at the end; the chat stays tidy. Nothing is changed until you confirm. A separate button **“Subagent stoppen”** (stop subagent) cancels only it; what it had by then stays.

> **Result:** Of five items, two had the wrong type: a dissertation was recorded as a journal article (“thesis”), a conference paper likewise (“conferencePaper”).

**Keep in mind:** With hundreds of items this takes minutes to hours, depending on how busy the server is.

## 5.14 Ask a book a substantive question

**You write (PDF chat, chapter 3):** *According to the book, what parts should a data management plan consist of? For each part, state what it is supposed to determine, and which example wordings the book gives for it. Support every statement with the page.*

So far the examples translated Zotero tasks into language. Here it is about the literature itself: you want to **work out a procedure from a long book** – here the structure of a data management plan together with model texts. This is a question to *one* book (Schlenz et al., 109 pages), hence the PDF chat in the side pane: it reads the whole book and gives page references.

1. Select the item and ask the question in the section “Chat mit PDF”. Phrase **a clear structure** (“for each part …”) and ask for **page references**.
2. The line above the answer shows what was read (here: full text, 109 pages).

# 6 Saving chats

## 6 What is volatile and what stays

**Chats are volatile.** A chat lives only in Zotero's working memory:

- It is kept as long as Zotero is running – also if you close the window, or switch to another item and come back. In the PDF chat there is one chat **per document**.
- It is gone after **“Neuer Chat”** (new chat) and after **restarting Zotero** (also after a crash or an update).

**PDF chat (side pane):**

- **“Verlauf als Notiz”** (chat as note) saves the whole conversation as a Zotero note on the item; the page references stay clickable links.
- The button **“Als Notiz”** (as note) on a single answer takes over only that answer.
- The chat can also be saved as a Markdown file.

**Chat over the library (chapter 4):** What the chat *did* stays in Zotero (imported items, changed metadata, collections, notes). To keep the **text of the conversation**:

1. **“⤓ Chat als .md”** (chat as .md) at the bottom of the window saves the whole conversation as a Markdown file: your questions, every tool call with its result and its items (links are kept) and the answers. You choose folder and file name.
2. If something should stay in Zotero, ask the model for a note: *Write the answer as a note on the item “…”* (see 5.9).

## 7 What to keep in mind

- **It is a language model, not a researcher.** It can summarize things wrongly or overlook hits. Page links and previews are there so that you check the source (5.14, 5.15).
- **Confirming means checking.** Look at changes in the preview, especially with many items. Zotero has no “undo” button for metadata; for mass changes make a backup first.
- **The model likes to do more than you asked.** Even with “change only the type” it sometimes adds title, DOI or pages (5.6, 5.13). Clear requests help (“only in this collection”, “at most 20 items”) but do not replace looking at the preview: untick there what you do not want.
- **Red lines in the tool block** are failed calls; the model usually repeats them by itself (5.9). If the error keeps coming back, phrase the request differently or smaller.
- **No index, no hits:** PDFs that ZotSeek or SeekBook have not indexed yet stay invisible to the search (chapter 2).
- **Data protection:** Text goes only to `ollama.ils.local`, the server in the server room, and only if it is entered under “Allowed remote hosts” – not to a cloud service.
- **Computing time:** The server is shared. Tasks over many items or whole books take minutes; you can keep working meanwhile.
- **Chats are volatile** (chapter 6): save what matters as a note.

## 8 If something does not work

- **No connection:** Are you in the ILS network? Is the address `https://ollama.ils.local` correct? Is `ollama.ils.local` entered under *Allowed remote hosts*? Then **Test connection** (chapter 2).
- **The model “just talks” and does nothing:** The chosen model is too small or not suited for Zotero functions – choose `qwen3_8_27b_128k`.
- **No hits from ZotSeek or SeekBook:** Is *AI Agent Access* switched on? Is the index finished (numbers in the settings, chapter 2)?
- **The answer breaks off or takes very long:** The server is busy. Cancel with the stop button and try again later or with a smaller question (fewer items, fewer pages).
- **Chat gone:** See chapter 6 – it was only in working memory.
- **Find out what happened:** Zotero → *Tools → Developer → Browser Console*, filter `[SeekChat`. Every step is logged there with its duration; this helps IT with troubleshooting.
